```python
import pandas as pd

# Loading the CSV file (file containing raw moisture values)
df = pd.read_csv(r"C:\Users\sambi\Desktop\TinyML\Soil_Quality_Data.csv")  # Change the path

# Conversion
df['Moisture (%)'] = (df['Soil Moisture'] - 300.0) / (1023.0 - 300.0) * 100.0
df['Moisture (%)'] = df['Moisture (%)'].clip(lower=0, upper=100)
df = df.drop(columns=['Soil Moisture'])
df = df[['Moisture (%)', 'Temperature', 'Air Humidity', 'Pump Data']]

# Saving to new CSV file
df.to_csv("Soil_Quality_Converted.csv", index=False)
print("Conversion complete. File saved as Soil_Quality_Converted.csv")
```
